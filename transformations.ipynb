```python
import requests
from collections import defaultdict

# URL api consultar
url_base = "https://pokeapi.co/api/v2/pokemon/"
url = "https://pokeapi.co/api/v2/pokemon-form"

my_pokemons = defaultdict(list)

#url_final = f"{url}/pokemon/{nombre_pokemon}"

response = requests.get(url)

pokemon_info = response.json()
results = pokemon_info.get('results', {})

if results:
    for pokemon in results:
        pokemon_name = pokemon['name']
        url_final = f"{url_base}/{pokemon_name}"
        response_detail = requests.get(url_final)
        results_detail = response_detail.json()
        my_pokemons["id"].append(results_detail["id"])
        my_pokemons["name"].append(results_detail["name"])
        my_pokemons["height"].append(results_detail["height"])
        my_pokemons["weight"].append(results_detail["weight"])
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

schema = StructType([
            StructField('id', IntegerType(), True), 
            StructField('name', StringType(), True),
            StructField('height', IntegerType(), True),
            StructField('weight', IntegerType(), True)
        ])
data = list(zip(*my_pokemons.values()))

df = spark.createDataFrame(data, schema = schema)

df.show()
```
